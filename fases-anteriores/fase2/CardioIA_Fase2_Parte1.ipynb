```python
# Importação das bibliotecas

import warnings
import pandas as pd
import unicodedata
from pathlib import Path
```

```python
# APRESENTAÇÃO DO CONTEÚDO NA LISTA
def ApresentaTXT(lista: dict):
    print("SEGUE APRESENTAÇÃO DOS CONTEÚDOS NA LISTA:\n")
    for conteudo in lista:
        print(conteudo)
```

```python
# APRESENTAÇÃO DO ARQUIVO CSV
def ApresentaCSV(lista: pd.DataFrame):
    print("SEGUE APRESENTAÇÃO DOS CONTEÚDOS NA LISTA:\n")
    print(lista.head())
```

```python
# LEITURA DO ARQUIVO TXT
def LeituraTXT(arq: str):
    try:
        erro = True
        conteudo = []

        with open(arq, 'r', encoding="utf-8") as arquivo:
            for linha in arquivo:
                conteudo.append(linha.strip())
        erro = False

    except FileNotFoundError:
        print("Arquivo não encontrado.")
    except PermissionError:
        print("Permissão negada para abrir o arquivo.")
    except UnicodeDecodeError:
        print("Erro de codificação ao ler o arquivo.")
    except Exception as e:
        print("Erro inesperado: {}".format(e))
    return conteudo, erro
```

```python
# LEITURA DO ARQUIVO CSV
def LeituraCSV(arq: pd.DataFrame):
    try:
        erro = True
        conteudo = pd.read_csv(arq)
        erro = False

    except FileNotFoundError:
        print("Arquivo não encontrado.")
    except FileNotFoundError:
        print("Arquivo não encontrado.")
    except PermissionError:
        print("Permissão negada para abrir o arquivo.")
    except pd.errors.EmptyDataError:
        print("Arquivo CSV está vazio.")
    except Exception as e:
        print(f"Ocorreu um erro inesperado: {e}")

    return conteudo, erro
```

```python
# NORMALIZAçÃO DO TEXTO
def Normalizar(frase: str) -> str:
    frase = frase.lower()

    # Remove acentos e caracteres especiais (ex: ç, ã, é)
    frase = ''.join(
        caracter for caracter in unicodedata.normalize('NFD', frase)
        if unicodedata.category(caracter) != 'Mn')

    return frase
```

```python
# VERIFICA SE UM TERMO ESTÁ CONTIDO NA FRASE
def contem(frase_norm: str, termo_norm: str) -> bool:

    # Se o termo estiver vazio
    if not termo_norm:
        return False

    # Verifica se o termo aparece dentro da frase
    if termo_norm in frase_norm:
        return True
    else:
        return False
```

```python
# LEITURA E APRESENTAÇÃO DO DOCUMENTO .CSV

documento_csv, erro_csv = LeituraCSV('mapa_conhecimento_cardio.csv')

if not erro_csv:
  ApresentaCSV(documento_csv)
```

## **↪ Processamento dos dados**

```python
# PREPARA ARQUIVO NORMATIZADO

mapa = documento_csv.copy()

if "Sintoma 1" in mapa.columns and "Sintoma 2" in mapa.columns:
    mapa["Sintoma 1"] = mapa["Sintoma 1"].astype(str).map(normaliza)
    mapa["Sintoma 2"] = mapa["Sintoma 2"].astype(str).map(normaliza)

    print(mapa.head())
else:
    print("As colunas esperadas 'Sintoma 1' e 'Sintoma 2' não foram encontradas no CSV.")
```

## **↪ Exibição dos resultados**

```python
# EXIBIÇÃO DOS RESULTADOS

df_result = pd.DataFrame(resultados)

# Salvamento em um arquivo externo .csv
out_path = Path("Diagnostico_frases_pacientes.csv")
df_result.to_csv(out_path, index=False, encoding="utf-8")

print("Arquivo salvo em: {}\n".format(out_path))
df_result.head()
```
